```python
%%sql
CREATE TABLE students(
    students_id INT,
    name char(50),
    age INT,
    grade char(1)
);
```

```python
%%sql
insert into students(students_id, name, age, grade)
values (1,'Shivpal', 21, 'A'),
       (2, 'Ram', 22, 'B')
```

```python
%%sql
select * from students;
```

```python
%%sql
select name from students where age = 21;
```

```python
%%sql
update students
set age = 24
where name = 'Ram';
select * from students;
```

```python
%%sql
delete from students
where name = 'Ram';
```
