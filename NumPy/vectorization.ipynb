```python
import numpy as np
```

ELEMENT WISE OPERATIONS 

```python
arr1 = np.array([1,2,3,4])
arr2 = np.array([5,6,7,8])
# res = arr1*arr2
# res = arr1+arr2
# res = arr1-arr2
res = arr1/arr2
print(res)
```


SCALER OPERATION

```python
res = arr1*3
print(res)
```


COMPARISION OPERATION

```python
res = arr1<arr2
print(res)
```


DOT PRODUCT

```python
matrix1 = np.array([[3,4,5],[3,5,1],[5,6,7]])
matrix2 = np.array([[2,3,4],[4,5,7],[4,4,3]])
print(matrix1*matrix2)
print(np.dot(matrix1,matrix2))
```


MATH FUNCTION

```python
a = np.array([0,np.pi/4,np.pi])
print(np.sin(a))
print(np.cos(a))
```
