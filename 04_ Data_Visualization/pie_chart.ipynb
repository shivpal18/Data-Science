```python
import matplotlib.pyplot as plt
```

```python
x= [10,20,30,40]
y=["C","C++","Java","Python"]
ex=[0.2,0.0,0.0,0.0]
c=["r","b","g","y"]
plt.pie(x,labels=y,explode=ex, colors=c, autopct="%0.1f%%", shadow=True, radius=0.8, labeldistance=1.1, startangle=90, textprops={"fontsize": 8}, counterclock=False, wedgeprops={"linewidth":2, "edgecolor":"r"}, rotatelabels=False)

plt.title("Pie chart")
plt.legend(loc=1)

plt.show()
```

```python
plt.pie([1])
plt.show()
```
