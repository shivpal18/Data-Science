```python
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns
```

```python
x=["A", "B", "C"]
y=[10,20,15]
plt.bar(x,y)
plt.show()
```

```python
plt.barh(x,y, color="green")
plt.show()
```

```python
df = pd.DataFrame({"cat":["A", "B", "C"], "val": [10,20,15]})
sns.barplot(x="cat", y="val", data=df)
```

```python
sns.barplot(x="val",y="cat", data=df, orient="h")
```
