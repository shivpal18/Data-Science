```python
import matplotlib.pyplot as plt
```

Add tet on the plot using annotate() function

```python
temp = [15,25,35,45,55]
visitors = [150, 186, 350, 500,900]

plt.scatter(temp, visitors)
plt.annotate("Highest Visitors", (55,900),(39,885), color='orange', fontsize=15, fontfamily='cursive')

plt.xlabel('Average Daily Temperature')
plt.ylabel("No of visitors")

plt.title("Minimum beach visitors")

plt.show()
```

Add text on the plot with arrow using annotate() function

```python
import numpy as np
```

```python
x=np.arange(0,10,0.2)
y=np.cos(x)

plt.plot(x,y, lw=2)
plt.annotate("max value", xy=(6.4, 1), xytext=(2, 0.95), arrowprops={})

plt.show()
```

Add annotation outside the plot

```python
x=[1,2,3,4]
y=[10,20,30,40]

plt.plot(x,y, marker='o')
plt.annotate('line chart', (0.5,25), (-1,25), arrowprops={}, annotation_clip=False)

plt.show()
```

Add arrow on the plot

```python
x=[1,2,3,4,5]
y=[3,6,9,12,15]

plt.plot(x,y)
plt.arrow(0.5, 1.75, 0.15, 0.5, head_width = 0.2, width = 0.1, ec = 'green', fc='orange')

plt.title('plot with an arrow')
plt.show()
```
