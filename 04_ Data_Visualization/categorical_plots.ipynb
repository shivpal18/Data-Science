```python
import matplotlib.pyplot as plt
import seaborn as sns
```

# Categorical Data Plots

Now let's discuss using seaborn to plot categorical data! There are a few main plot types for this:

1) boxplot
2) voilinplot
3) stripplot
4) swarmplot
5) barplot
6) countplot

Let's go through examples of each

```python
df = sns.load_dataset('tips')
df
```

```python
sns.countplot(x = df['sex'])
```

```python
sns.countplot(x = df['sex'], hue= df['smoker'])
```

```python
sns.barplot(x = df['sex'], y = df['total_bill'])
```

You can change the estimator object to your own function, that convets a vector to a scalar Mean (default): np.mean Median: np.median Sum: np.sum Count; len(to count the number of observations) Standard Deviation: np.std Minimum: np.min Maxium: np.max

```python
import numpy as np
sns.barplot(x=df['sex'], y = df['tip'], estimator=np.sum)
```

```python
sns.boxplot(x='tip', y='day', data=df, palette='rainbow')
```

# boxplot and voilinplot
boxplots and voilinplot are used to shown the distribution of categorical data. A box plot (or box-and-whisker-plot) shows the distribution of quantitative of the dataset while the whiskers extend to show the rest of the distribution, except for points that are determined to be "outliers" using a method that is a function of the inter-quartile range.

```python
sns.violinplot(x='day', y='total_bill', data = df)
```

# Stripplot and swarmplot
The stripplot will draw a scatterplot where one variable is categorical. A strip plot can be drawn on its own, but it is also a good complment to a box or violin plot in cases where you want to show all observations along with some representation of the underlying distribution.

The swarmplot is similar to stripplot(), but the points are adjusted (only along the categorical axis) so that they don't overlap. This gives a better representation of the distribution of values, although it does not scale as well to large numbers of observations (both in terms of the ability to show all the points and in terms of the computation needed to arraange them).

```python
sns.stripplot(x = 'day', y = 'total_bill', data = df)
```

```python
sns.swarmplot(x='day', y='total_bill', data = df)
```

```python
sns.violinplot(x='tip', y='day', data = df)
sns.swarmplot(x='tip', y='day', data = df)
```
