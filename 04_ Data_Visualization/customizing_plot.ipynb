```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
x = np.arange(10)
```

```python
y=x*2
```

```python
x
```

```python
y
```

```python
plot = plt.figure()
axplot = plot.add_axes([0,0,1,1])

axplot.plot(x,y)
axplot.set_xlabel('X axis')
axplot.set_ylabel('Y axis')
axplot.set_title('Title')
plt.show()
```

```python
plot = plt.figure()
axplot = plot.add_axes([0,0,1,1])

axplot.plot(x,y, label = 'x')
axplot.plot(x*3,y, label = 'x*3')
axplot.legend(loc = 1)
axplot.set_xlabel('X axis')
axplot.set_ylabel('Y axis')
axplot.set_title('Title')
plt.show()
```

```python
fig = plt.figure()
ax = fig.add_axes([0,0,1,1])
ax.plot(x,x,color='blue')
ax.plot(x,x*1.5,color='blue')
ax.plot(x,x*2,color='#D31ED4')
ax.plot(x,x*2.5,color='#D31ED4', alpha = 0.3)
ax.plot(x,x*3,color='black', linewidth=1)
ax.plot(x,x*3.5,color='black', lw=1)
ax.plot(x,x*4,color='black', lw=3, linestyle = '-.')
ax.plot(x,x*4.5,color='red', lw=3, ls=':')
ax.plot(x,x*5,color='red', lw=1, marker='o')
ax.plot(x,x*5.5,color='red', lw=1, marker='s', markersize = 10)
ax.plot(x,x*6,color='red', lw=1, marker='s', markersize = 10, markerfacecolor='green')
```
