```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
data_set = sns.load_dataset('taxis')
print(data_set.head(3))
```

```python
sns.set(style="whitegrid")
pt_1 = sns.FacetGrid(data=data_set, col="color", hue="payment")
pt_1.map(sns.scatterplot, "distance", "total", alpha = 0.8)
pt_1.add_legend()
plt.show()
```

```python
sns.set(style="darkgrid")
pt_2 = sns.FacetGrid(data=data_set, col="color", col_wrap=1)
pt_2.map(sns.histplot, "tolls")
plt.show()
```
