```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
data_set = sns.load_dataset('taxis')
print(data_set.head(3))
```

```python
sns.set(style="whitegrid")
pt_1 = sns.FacetGrid(data=data_set, col="color", hue="payment")
pt_1.map(sns.scatterplot, "distance", "total", alpha = 0.8)
pt_1.add_legend()
plt.show()
```

```python
sns.set(style="darkgrid")
pt_2 = sns.FacetGrid(data=data_set, col="color", col_wrap=1)
pt_2.map(sns.histplot, "tolls")
plt.show()
```

```python
sns.set(style="ticks")
pt_3 = sns.FacetGrid(data=data_set, row = "color", col="payment", margin_titles=True)
pt_3.map(sns.regplot, "fare", "tip", color = "r", fit_reg = False, x_jitter = 0.2)
plt.show()
```

```python
sns.set(style = "white")
pt_4 = sns.FacetGrid(data=data_set, row = "passengers", height=4.5, aspect=0.9)
pt_4.map(sns.barplot, "color", "tip", order = ["yellow", "green"])
plt.show()
```

```python
sns.set(style='ticks')
fig, axes = plt.subplots(2,3, figsize=(18,10))
fig.suptitle('Multi plot grid')
data_set = sns.load_dataset("taxis")

sns.barplot(ax=axes[0,0], data=data_set, x='color', y='total')
sns.boxplot(ax=axes[0,1], data=data_set, x='color', y='fare')
sns.scatterplot(ax=axes[0,2], data=data_set, x='color', y='tip')
sns.violinplot(ax=axes[1,0], data=data_set, x='color', y='tolls')
sns.barplot(ax=axes[1,1], data=data_set, x='color', y='passengers')
sns.boxplot(ax=axes[1,2], data=data_set, x='color', y='distance')
```
