```python
import matplotlib.pyplot as plt
```

```python
x=[1,2,3,4,5]
y=[1,2,1,2,1]

yerrormin = [0.1,0.5,0.9,0.1,0.9]
yerrormax = [0.2,0.4,0.6,0.4,0.2]
xerror = 0.5

yerror = [yerrormin, yerrormax]

# plt.plot(x,y,'o')
plt.errorbar(x,y,yerr=yerror,xerr=xerror, fmt='o', ecolor='black', elinewidth=2, capsize=5, capthick=2, barsabove=False, errorevery=1, lolims=True, xlolims=True, xuplims=False)

plt.show()
```
