```python
import numpy as np
a = np.array([40,10,30,20])
print(np.sort(a))
```

```python
a = np.array([[3,2,1],[6,5,4]])
print(np.sort(a))
```

```python
a = np.array([40,10,30,20])
print(np.argsort(a))
```

```python
a = np.array([10,20,30,40,50])
print(np.where(a==30))
```

```python
a = np.array([10,15,20,25,30])
print(np.where(a%2==0))
```

```python
a = np.array([10,20,30,40,50])
print(np.searchsorted(a, 35))
```

```python
a = np.array([[3,1,2],[6,4,5]])
print(np.sort(a, axis=0))
print(np.sort(a, axis=1))
```
